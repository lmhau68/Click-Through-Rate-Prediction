```python
import os
import pandas as pd
from sklearn.preprocessing import StandardScaler

# 1. Đường dẫn file dữ liệu thô
raw_data_path = '../../data/raw/ad_10000records.csv'
if not os.path.exists(raw_data_path):
    raw_data_path = 'data/raw/ad_10000records.csv'

df = pd.read_csv(raw_data_path)
print(f"Kích thước ban đầu: {df.shape}")

# 2. Xử lý Categorical: Chuyển Gender thành dạng số (Male: 1, Female: 0)
df['Gender'] = df['Gender'].map({'Male': 1, 'Female': 0})

# 3. Loại bỏ các cột phân mảnh cao, không dùng cho baseline
cols_to_drop = ['Ad Topic Line', 'City', 'Country', 'Timestamp']
df_cleaned = df.drop(columns=cols_to_drop)

# 4. Chuẩn hóa đặc trưng số (Feature Scaling)
num_cols = ['Daily Time Spent on Site', 'Age', 'Area Income', 'Daily Internet Usage']
scaler = StandardScaler()
df_cleaned[num_cols] = scaler.fit_transform(df_cleaned[num_cols])

# 5. Xuất file data sạch cho Thành viên 2
output_dir = '../../data/processed'
output_path = os.path.join(output_dir, 'ctr_cleaned.csv')

if not os.path.exists(output_dir):
    output_path = 'data/processed/ctr_cleaned.csv'

df_cleaned.to_csv(output_path, index=False)
print(f"Xử lý thành công! File đã lưu tại: {output_path}")
print(df_cleaned.head())
```
